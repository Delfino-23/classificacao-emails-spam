```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Localização dos resultados
pasta = Path("../resultados")

# Arquivos gerados pelos três modelos
arquivos = [
    pasta / "naive_bayes" / "metricas.csv",
    pasta / "regressao_logistica" / "metricas.csv",
    pasta / "svm" / "metricas.csv"
]

# Lê e reúne os resultados
tabelas = [pd.read_csv(arquivo) for arquivo in arquivos]

comparacao = pd.concat(tabelas, ignore_index=True)

# Exibe a comparação
display(comparacao)
```

```python

metricas_comparacao = comparacao.melt(
    id_vars="Modelo",
    value_vars=[
        "Acuracia",
        "Precisao_Spam",
        "Recall_Spam",
        "F1_Spam"
    ],
    var_name="Métrica",
    value_name="Resultado"
)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=metricas_comparacao,
    x="Métrica",
    y="Resultado",
    hue="Modelo"
)

plt.title("Comparação de desempenho dos modelos")
plt.ylabel("Resultado")
plt.xlabel("Métrica")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```

```python

comparacao.to_csv(
    "comparacao_modelos.csv",
    index=False
)

print("Comparação geral salva com sucesso!")
```
